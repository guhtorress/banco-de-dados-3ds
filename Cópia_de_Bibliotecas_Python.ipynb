```python
# Importa as bibliotecas necessárias do Python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#1. Cria um DataFrame (tabela) com as notas

dados = {
'Aluno': ['João', 'Maria', 'Ana', 'Pedro'],
'Matemática': [90, 85, 88, 92],
'História': [88, 79, 85, 80]
}
df = pd.DataFrame(dados)
df
```

```python
#2. Calcula a média (apenas colunas numéricas)

df['Média'] = df[['Matemática', 'História']].mean(axis=1)
print('Tabela de Dados:\n', df)
```

```python
#3. Cria um gráfico de barras

df.plot(x='Aluno', y=['Matemática', 'História', 'Média'], kind='bar', figsize=(8,5))
plt.title('Desempenho dos Alunos')
plt.ylabel('Notas')
plt.show()
```
